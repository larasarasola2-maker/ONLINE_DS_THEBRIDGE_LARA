## Ejercicio 2
Haz un programa que detecte todos los duplicados de un elemento en una lista:
* Utiliza una variable duplicado para asignarle el valor del elemento del que queremos encontrar todos sus duplicados.
* El programa debe devolver todos los índices de los elementos duplicados.
* Aplícalo para encontrar los duplicados de "un", "es" y "binario" de:  

>```ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]```


`````python

ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]
dic_duplicados = {}
#Pasamos todos los elementos a minúsculas
for elemento in ejer_3:
  if (type(elemento) == str):
    ejer_3[ejer_3.index(elemento)] = elemento.lower()

#Hacemos un set
ejer_3_set = set(ejer_3)
print(ejer_3)
print(ejer_3_set)

#Metemos los elementos del set como claves de un diccionario
for elemento in ejer_3_set:
  dic_duplicados[elemento] = []

#Comparamos la lista original con el set para saber los elementos que se repiten
#Creo un contador para llevar la cuenta de en qué índice de la lista original estoy
for elemento in ejer_3_set:
  contador = -1
  for elemento2 in ejer_3:
    contador += 1
    if elemento == elemento2:
      dic_duplicados[elemento].append(contador)

print(dic_duplicados)
#Paso la lógica a una función para usarla en el siguiente ejer
def contador_duplicados(lista):
  dic = {}

  for elemento in lista:
    if (type(elemento) == str):
      lista[lista.index(elemento)] = elemento.lower()

  #Hacemos un set
  lista_set = set(lista)


  #Metemos los elementos del set como claves de un diccionario
  for elemento in lista_set:
    dic[elemento] = []

#Comparamos la lista original con el set para saber los elementos que se repiten
#Creo un contador para llevar la cuenta de en qué índice de la lista original estoy
  for elemento in lista_set:
    contador = -1
    for elemento2 in lista:
      contador += 1
      if elemento == elemento2:
        dic[elemento].append(contador)
  return dic
contador_duplicados(ejer_3)
`````
