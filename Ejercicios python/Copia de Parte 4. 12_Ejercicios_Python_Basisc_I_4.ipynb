## Ejercicio 1
Para este ejercicio vamos a poner en práctica [las funciones built in](https://docs.python.org/3/library/functions.html).

1. Calcula el máximo de la lista: [4, 6, 8, -1]
2. Suma todos los elementos de la lista anterior
3. Redondea este float a 3 dígitos decimales: 63.451256965
4. Valor absoluto de: -74

```python
lista = [4, 6, 8, -1]
print(max(lista))
print(sum(lista))

numero_real = 63.451256965
print(round(numero_real, 3))

numero_entero = -74
print(abs(numero_entero))
```

## Ejercicio 2
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

```python
cadena = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
print(cadena.upper())

print(cadena.lower())

print(cadena.title())

print(cadena.split())

print(cadena.replace(",", ";"))


print(cadena.replace("a", ""))
```

## Ejercicio 3
1. Crea una lista con 3 elementos numéricos
2. Añade un cuarto elemento
3. Calcula la suma de todos
4. Elimina el segundo elemento de la lista
5. Añade otro elemento en la posicion 3 de la lista
6. Crea otra lista con 4 elementos y concaténala a la que ya tenías.
7. Ordena la lista de menor a mayor.
8. BONUS: Ordena la lista de mayor a menor.  
(*PISTA: Para hacer 8, tendrás que buscar en la información del método...*)  
  
  
NOTA: Pinta por pantalla el resultado de cada paso.

```python
lista = [1,2,3]
lista.append(4)

print(lista)

print(sum(lista))

lista.pop(1)
print(lista)

lista.insert(2,100)
print(lista)

lista_2 = [5,6,7,8]
lista_suma = lista + lista_2
print(lista_suma)

lista_suma.sort()
print(lista_suma)

lista_suma.sort(reverse=True)
print(lista_suma)
```
